### Una lista es una secuencia

Como un string, una **lista** es una secuencia de valores. La diferencia: en un string los valores son caracteres; en una lista pueden ser de cualquier tipo. Cada valor es un **elemento**, y la lista se escribe entre corchetes.

### Porciones de lista

Una **porción** (en inglés, *slice*) es un trozo de la lista. Se escribe con dos índices separados por dos puntos, `lista[inicio:fin]`, y el resultado es **otra lista**. Funciona igual que con los strings, pero es lo que más cuesta del capítulo, así que vamos despacio.

El truco para no equivocarse es que, en una porción, los números no señalan las cajas, sino **los huecos entre las cajas**. `temperaturas[1:3]` quiere decir «corta en el hueco 1 y en el hueco 3, y quédate con lo que queda en medio».

![Pizarra: los índices de porción señalan los huecos](https://raw.githubusercontent.com/maximofernandezriera/000-thinkpython/main/pizarra-porciones.png)

```python
temperaturas = [-63, -58, -71, -49, -66]
temperaturas[1:3]
```

Dos elementos, los de las posiciones 1 y 2. El de la posición 3 (-49) se queda fuera: **`inicio` entra, `fin` no**. Una comprobación rápida es que una porción `[a:b]` tiene siempre `b - a` elementos, si los dos índices están dentro de la lista:

```python
len(temperaturas[1:3]), 3 - 1
```

#### Omitir el inicio o el fin

Si falta el primer número, la porción empieza por el principio; si falta el segundo, llega hasta el final. Predecid antes de ejecutar cada celda:

```python
temperaturas[:2]     # los dos primeros sols
```

```python
temperaturas[2:]     # del sol de la posición 2 en adelante
```

`[:2]` y `[2:]` cortan en **el mismo hueco**, así que se reparten la lista sin repetir ni dejarse ningún elemento:

```python
temperaturas[:2] + temperaturas[2:]
```

Y sin ninguno de los dos números, `[:]`, la porción es la lista entera: una **copia**. Nos hará falta en la sección de aliasing.

```python
copia = temperaturas[:]
copia
```

En PythonTutor se ve que cada porción es una lista **nueva**, con su propia flecha, y que la original no cambia:

#### Elemento frente a porción

Esta es la confusión más habitual. Con un índice se obtiene **un elemento**; con una porción, **una lista**, aunque solo tenga un elemento. Predecid qué imprime cada línea:

```python
print(temperaturas[1], type(temperaturas[1]))
print(temperaturas[1:2], type(temperaturas[1:2]))
```

¿Por qué importa? Porque con el elemento se puede operar como con cualquier número, y con la porción no:

```python
temperaturas[1] + 10
```

```python
temperaturas[1:2] + 10
```

`TypeError`: el `+` de las listas sirve para concatenar listas, y `10` no es una lista. En PythonTutor se ve la diferencia: `elemento` es un número suelto y `porcion` es una lista de una sola caja.

### Recorrer una lista con un bucle

El `for` del capítulo 7 recorre una lista exactamente igual que recorría un string: un elemento por vuelta.

```python
for persona in tripulacion:
    print("Traje presurizado:", persona)
```

Y el patrón *inicializar, recorrer, actualizar* sigue funcionando. Sin usar `min`: ¿cuál ha sido el sol más frío? Guardamos el mínimo visto hasta ahora y lo actualizamos cuando aparece uno menor.

```python
def mas_frio(temps):
    minimo = temps[0]          # inicializar con el primero
    for t in temps:
        if t < minimo:         # actualizar si encontramos uno menor
            minimo = t
    return minimo

mas_frio(temperaturas)
```

Seguid el bucle paso a paso: fijaos en cómo `t` toma cada temperatura y `minimo` solo cambia cuando aparece una más baja (-63 → -71).

El mismo patrón sirve para **construir** una lista nueva: empezar con `[]` y hacer `append` dentro del bucle. Por ejemplo, convertir las temperaturas a kelvin:

```python
def a_kelvin(temps):
    resultado = []
    for t in temps:
        resultado.append(round(t + 273.15, 2))   # round: dos decimales
    return resultado

a_kelvin(temperaturas)
```

### Objetos y valores

Ahora viene la parte del capítulo que más errores evita en la vida real. ¿Qué creéis que da cada una de estas dos celdas?

```python
a = [1, 2, 3]
b = [1, 2, 3]
a == b
```

`==` compara **valores**: las dos listas tienen los mismos elementos, así que son **equivalentes**. `is` pregunta si son **el mismo objeto**, y no lo son: hemos creado dos listas distintas que casualmente contienen lo mismo. Equivalentes, pero no **idénticas**.

### Aliasing

Ahora en lugar de crear dos listas asignamos una variable a otra. Antes de ejecutar: después de la tercera línea, ¿qué vale `equipo_a`?

```python
equipo_a = ["Ada", "Mae"]
equipo_b = equipo_a          # NO copia la lista: crea otro nombre para la misma
equipo_b.append("Yuri")
equipo_a
```

```python
equipo_a is equipo_b
```

`equipo_b = equipo_a` no crea una lista nueva: crea **otro nombre** para la misma lista. Cada asociación entre una variable y un objeto es una **referencia**; cuando un objeto tiene dos nombres decimos que tiene un **alias**. Como la lista es mutable, lo que se cambia por un nombre se ve por el otro.

El esquema de la pizarra:

Con objetos inmutables, como los strings, el aliasing no da problemas: nadie puede cambiar el objeto compartido.

### La lista de palabras del libro

En el capítulo 7 leíamos `words.txt` entero cada vez que buscábamos algo. Es mucho más eficiente leerlo **una vez** y guardar las palabras en una lista.

```python
from os.path import basename, exists

def descargar(url):
    archivo = basename(url)
    if not exists(archivo):
        from urllib.request import urlretrieve
        urlretrieve(url, archivo)
    return archivo

descargar('https://raw.githubusercontent.com/AllenDowney/ThinkPython/v3/words.txt')
```

```python
# versión 1: inicializar con [] y append en cada vuelta
palabras = []
for linea in open('words.txt'):
    palabras.append(linea.strip())

len(palabras)
```

```python
# versión 2: leer todo el archivo como un string y partirlo con split
palabras = open('words.txt').read().split()
len(palabras)
```

### Depuración: la mayoría de métodos de lista devuelven None

```python
cola = ["basalto", "olivino", "hematita"]
cola = cola.remove("olivino")    # ¡MAL!
print(cola)
```

`AttributeError: 'NoneType' object has no attribute 'append'`. `remove` modificó la lista y devolvió `None`, y al asignarlo hemos perdido la lista. Cuando veáis un `NoneType` donde esperabais una lista, buscad un método de lista cuyo resultado se haya asignado. Lo correcto es llamar al método sin asignar:

```python
cola = ["basalto", "olivino", "hematita"]
cola.remove("olivino")
cola
```

Comparad las dos versiones en PythonTutor: en la mala, después de la asignación `cola` deja de apuntar a la lista y pasa a valer `None`; en la buena, la lista sigue ahí, con un elemento menos.

- [**Find the smallest integer in the array**](https://www.codewars.com/kata/55a2d7ebe362935a210000b2): el mínimo de una lista. Se resuelve con `min`... y luego intentadlo con el patrón de `mas_frio`.
- [**Sum of positive**](https://www.codewars.com/kata/5715eaedb436cf5606000381): sumar solo los positivos. Acumulador + `if` dentro del `for`.
- [**Beginner - Lost Without a Map**](https://www.codewars.com/kata/57f781872e3d8ca2a000007e): devolver una lista nueva con cada valor multiplicado por 2. Es el patrón de `a_kelvin`.
- [**Reversed Words**](https://www.codewars.com/kata/51c8991dee245d7ddf00000e): dar la vuelta al orden de las palabras de una frase. `split` + `join`, igual que el ejercicio 3.

## Cierre

1. Una lista es una secuencia de valores de cualquier tipo, entre corchetes; se indexa desde 0 y admite porciones como los strings.
2. Las listas son **mutables**: se cambian con `lista[i] = ...`, `append`, `extend`, `pop` y `remove`. Casi todos esos métodos devuelven `None`, así que nunca hagáis `t = t.remove(x)`.
3. `split` convierte texto en lista y `join` lista en texto; `sorted` devuelve una lista nueva ordenada.
4. `==` compara valores e `is` identidad. `b = a` no copia: crea un **alias**. Para copiar, `a[:]` o `list(a)`.
5. Pasar una lista a una función también crea un alias: si la función la modifica, el cambio se queda.

Próxima parada: capítulo 10, diccionarios. Las listas encuentran las cosas por su posición; los diccionarios, por su nombre.
